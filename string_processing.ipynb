```python
import pandas as pd
```

```python
df=pd.read_csv("/content/drive/MyDrive/Copy of string_processing_dataset.csv")
```

```python
df
```

```python
df["email"]=df["email"].str.lower()
df
```

```python
df["city"]=df["city"].str.upper()
df
```

```python
df["customer_name"]=df["customer_name"].str.title()
df
```

```python
df["name_length"]=df["customer_name"].str.len()
df
```

```python
df
```

```python
#regex=false tells pandas to treat the value as normal text ad not as a spl pattern
df["purchase_amount"]=df["purchase_amount"].str.replace("Rs ","", regex= False)
df
```

```python
df["purchase_amount"]=df["purchase_amount"].str.replace("$","", regex= False)
df
```

```python
df["purchase_amount"]=df["purchase_amount"].str.replace(",","", regex= False)
df
```

```python
df["is_mail"]=df["email"].str.contains("gmail")
df
```

```python
df["startswithv"]=df["city"].str.startswith("v")
df
```

```python
df["endswith"]=df["email"].str.endswith(".com")
df
```

```python
df["emailparts"]=df["email"].str.split("@").str[0]
df
```

```python
df["emailpartscombined"]=df["email"].str.split("@")
df
```

```python
df["city_column"]=df["city"].str[:3]
df
```

```python
df["is_numeric"]=df["phone"].astype(str).str.isnumeric()
df
```

```python
df["product_code"]=df["product_code"].str.removeprefix("PRD-")
df
```
