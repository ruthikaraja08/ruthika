```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/student_data_cleaning.csv")
df
```

```python
df.info()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
df
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates(df)
```

```python
df
```

```python
df["City"]=df["City"].replace(
    {
        "HYD":"Hyderabad",
        "hyd":"Hyderabad"
    }
)
df
```

```python
df.columns=(
    df.columns.str.lower().str.replace(" ","_")
)
df
```

```python
df["join_date"]=pd.to_datetime(
    df["join_date"],
    format="mixed"
)
df
```
