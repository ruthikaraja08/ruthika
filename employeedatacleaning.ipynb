```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/employee_data.csv")
df
```

```python
df.info()
```

```python
df["Age"]=df["Age"].replace(
    {
        "NaN":"Average Age"
    }
)
df
```

```python
df["Employee Name"]=df["Employee Name"].replace(
    {
        "NaN":"Unknown"
    }
)
df
```

```python
df.duplicated().sum()
```

```python
df.columns=(
    df.columns.str.lower().str.replace(" ","_")
)
df
```

```python
df["department"]=df["department"].replace(
    {
        "IT":"Information Technology",
        "it":"Information Technology"
    }
)
df
```

```python
df["join_date"]=pd.to_datetime(
    df["join_date"],
    format="mixed"
)
df
```
